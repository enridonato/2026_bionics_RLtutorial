## 1. Variables and numbers
A variable is a name attached to a value. `#` starts a comment: Python ignores it.

````python
length = 0.3          # metres
angle = 45            # degrees
print("length =", length, "m")
print("half length:", length / 2)
print("type of length:", type(length), "- type of angle:", type(angle))
````

## 2. Lists, loops and conditions
A **list** holds several values. A **for** loop repeats something for each element. **Indentation** (4 spaces) defines what is inside the loop.

````python
joint_angles = [0.0, 0.5, -1.2]     # a list with 3 numbers
print("first element:", joint_angles[0])   # Python counts from 0!
print("number of elements:", len(joint_angles))

for i, a in enumerate(joint_angles):
    if a < 0:
        print(f"joint {i}: {a} rad (negative)")
    else:
        print(f"joint {i}: {a} rad")
````

## 3. Functions
A function packs some code under a name, takes **inputs** (arguments) and **returns** an output. Here: the forward kinematics of a robot with **one** link of length `L`.

````python
import math

def fk_one_link(q, L=1.0):
    """Position (x, y) of the tip of a 1-link arm rotated by q radians."""
    x = L * math.cos(q)
    y = L * math.sin(q)
    return x, y

print(fk_one_link(0.0))            # arm along x
print(fk_one_link(math.pi / 2))    # arm along y
print(fk_one_link(math.pi / 2, L=0.5))
````

## 4. NumPy: computing with whole arrays at once
Robots and neural networks work with **vectors and matrices**. NumPy (`np`) handles them efficiently.
Rule of thumb: *avoid loops, write the operation on the whole array*.

````python
import numpy as np

q = np.array([0.0, 0.5, -1.2])        # a vector (1D array)
print("q        =", q)
print("2 * q    =", 2 * q)            # element-wise
print("sin(q)   =", np.sin(q))
print("shape    =", q.shape)

A = np.array([[1.0, 2.0],
              [3.0, 4.0]])            # a 2x2 matrix
v = np.array([1.0, 1.0])
print("A @ v    =", A @ v)            # @ = matrix-vector product
print("A.T      =\n", A.T)            # transpose
````

### Many samples at once
A dataset is a matrix: **one row per sample**. Here 1000 random angles and the corresponding tip positions, *without any loop*.

````python
rng = np.random.default_rng(0)                 # random generator (seed 0 -> reproducible)
qs = rng.uniform(-np.pi, np.pi, size=1000)     # 1000 random angles
xy = np.stack([np.cos(qs), np.sin(qs)], axis=1)  # shape (1000, 2)
print("qs.shape =", qs.shape, " xy.shape =", xy.shape)
print("first 3 rows:\n", xy[:3])
print("mean of each column:", xy.mean(axis=0))
````

## 5. Plots with Matplotlib

````python
import matplotlib.pyplot as plt

plt.figure(figsize=(4, 4))
plt.scatter(xy[:, 0], xy[:, 1], s=3, c=qs, cmap="twilight")
plt.colorbar(label="q [rad]")
plt.axis("equal"); plt.xlabel("x [m]"); plt.ylabel("y [m]")
plt.title("Workspace of a 1-link arm")
plt.show()
````

```python
def fk_two_links(q1, q2, L1=1.0, L2=0.8):
    x = L1 * np.cos(q1) + L2 * np.cos(q1 + q2)
    y = L1 * np.sin(q1) + L2 * np.sin(q1 + q2)
    return x, y
```
A ring with inner radius $L_1 - L_2 = 0.2$ and outer radius $L_1 + L_2 = 1.8$. Every point strictly inside the ring
is reached with **two** configurations: *elbow up* and *elbow down*. Keep this in mind: it is the key of lecture 3!
